# 04 — Model Training (PyTorch CNN-LSTM)

Loads processed splits from `data/processed/`, trains `CNNLSTMClassifier`, checkpoints the best weights to `models/cnn_lstm_best.pt`, and saves learning curves.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src import load_config
from src.data import load_processed_splits
from src.training import train_model

cfg = load_config(PROJECT_ROOT / "configs" / "config.yaml")
cfg["training"]["checkpoint_path"] = str(PROJECT_ROOT / cfg["training"]["checkpoint_path"])
cfg["results"]["plots_dir"] = str(PROJECT_ROOT / cfg["results"]["plots_dir"])
cfg["results"]["metrics_dir"] = str(PROJECT_ROOT / cfg["results"]["metrics_dir"])

splits = load_processed_splits(PROJECT_ROOT / cfg["data"]["processed_dir"])
model, history = train_model(
    splits["X_train"],
    splits["y_train"],
    splits["X_val"],
    splits["y_val"],
    cfg=cfg,
    verbose=True,
)
print(model)
```
