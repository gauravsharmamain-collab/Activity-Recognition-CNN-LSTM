# 01 — Data Collection (WISDM Dataset)

Loads raw tri-axial sensor recordings from the WISDM Smartphone and Smartwatch Activity and Biometrics dataset using `configs/config.yaml`.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src import load_config
from src.data import load_wisdm_raw, resolve_raw_data_dir

cfg = load_config(PROJECT_ROOT / "configs" / "config.yaml")
raw_dir = resolve_raw_data_dir(cfg["data"]["raw_dir"], PROJECT_ROOT / cfg["data"]["fallback_raw_dir"])
print("Resolved WISDM raw directory:", raw_dir)
```

```python
raw_df = load_wisdm_raw(
    raw_dir=raw_dir,
    devices=cfg["data"]["devices"],
    sensors=cfg["data"]["sensors"],
)
print("Loaded raw shape:", raw_df.shape)
print("Unique users:", raw_df["user"].nunique())
raw_df.head()
```
